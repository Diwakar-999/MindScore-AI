```python
top_countries = df['Country'].value_counts().index[:10].tolist()
```

```python
def group_countries(country):
  if country in top_countries:
    return country
  else:
    return 'Other'
```

```python
df['Grouped_countries'] = df['Country'].apply(group_countries)
```

```python
from sklearn.model_selection import train_test_split

skewed_col = ['Study_Hours']
numerical_cols = ['Age',	'Avg_Daily_Usage_Hours',	'Daily_Unlocks', 'Physical_Activity_Hours', 'Sleep_Hours_Per_Night']
ordinal_col = ['Stress_Level']
nominal_categorical_cols = ['Gender',	'Most_Used_Platform', 'Grouped_countries','Purpose_Of_Use','Academic_Level']

feature_cols = skewed_col + numerical_cols + ordinal_col + nominal_categorical_cols

X = df[feature_cols]
y = df['Mental_Health_Score']

x_train,x_test,y_train,y_test = train_test_split(X, y, test_size = 0.3, random_state = 42)
```

```python
df['Stress_Level'].value_counts()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,OrdinalEncoder,OneHotEncoder
from sklearn.compose import ColumnTransformer

# skewed_col
skewed_pipeline = Pipeline(steps=[
    ('log_transformation',FunctionTransformer(np.log1p)),
    ('Scale',StandardScaler())
])

# Other numerical features
numerical_pipeline = Pipeline(steps=[
    ('Scale',StandardScaler())
])

# Ordinal columns
ordinal_pipeline = Pipeline(steps = [
    ('OrdinalEncode',OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))
])

# Nominal categorical columns
nominal_pipeline = Pipeline(steps = [
    ('OneHotEncode',OneHotEncoder(handle_unknown='ignore'))
])

# (name of pipeline , Which pipeline , which columns)
columnTranformer = ColumnTransformer(transformers=[
    ('Skewed_pipeline',skewed_pipeline,skewed_col),
    ('numerical_pipeline',numerical_pipeline,numerical_cols),
    ('ordinal_pipeline',ordinal_pipeline,ordinal_col),
    ('nominal_pipeline',nominal_pipeline,nominal_categorical_cols)
])
columnTranformer
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

lr_pipeline = Pipeline(steps=[
    ('columnTransformer',columnTranformer),
    ('lrRegressor',LinearRegression())
])

lr_pipeline.fit(x_train,y_train)
lr_pred = lr_pipeline.predict(x_test)
lr_pred_train = lr_pipeline.predict(x_train)


print('Training metrices :- ')
print('R2 Score : ',r2_score(y_train,lr_pred_train))
print('MAE : ',mean_absolute_error(y_train,lr_pred_train))
print('MSE : ',mean_squared_error(y_train,lr_pred_train))
print("\n")
print('Testing metrices :- ')
print('R2 Score : ',r2_score(y_test,lr_pred))
print('MAE : ',mean_absolute_error(y_test,lr_pred))
print('MSE : ',mean_squared_error(y_test,lr_pred))
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_pipeline = Pipeline(steps=[
    ('columnTransformer',columnTranformer),
    ('rfRegressor',RandomForestRegressor(n_estimators= 249, max_depth= 46, min_samples_split= 3, min_samples_leaf= 1,random_state=42))
])

rf_pipeline.fit(x_train,y_train)
rf_pred = rf_pipeline.predict(x_test)
rf_pred_train = rf_pipeline.predict(x_train)

print('Training metrices :- ')
print('R2 Score : ',r2_score(y_train,rf_pred_train))
print('MAE : ',mean_absolute_error(y_train,rf_pred_train))
print('MSE : ',mean_squared_error(y_train,rf_pred_train))
print("\n")
print('Testing metrices :- ')
print('R2 Score : ',r2_score(y_test,rf_pred))
print('MAE : ',mean_absolute_error(y_test,rf_pred))
print('MSE : ',mean_squared_error(y_test,rf_pred))
```

```python
# It is the best model to give better accuracy amoung all three

import xgboost as xgb

xgb_pipeline = Pipeline(steps=[
    ('columnTransformer',columnTranformer),
    ('xgbRegressor',xgb.XGBRegressor(
        n_estimators = 2555,
        max_depth = 9,
        learning_rate = 0.039249201012217225,
        subsample = 0.8940728731653981,
        colsample_bytree = 0.5901009521371252,
        min_child_weight = 3,
        reg_alpha = 0.00010690391686694115,
        reg_lambda = 0.00346339928712085,
        gamma = 9.073443945710458e-05
    ))
])

xgb_pipeline.fit(x_train,y_train)
xgb_pred = xgb_pipeline.predict(x_test)
xgb_pred_train = xgb_pipeline.predict(x_train)

print('Training metrices :- ')
print('R2 Score : ',r2_score(y_train,xgb_pred_train))
print('MAE : ',mean_absolute_error(y_train,xgb_pred_train))
print('MSE : ',mean_squared_error(y_train,xgb_pred_train))
print("\n")
print('Testing metrices :- ')
print('R2 Score : ',r2_score(y_test,xgb_pred))
print('MAE : ',mean_absolute_error(y_test,xgb_pred))
print('MSE : ',mean_squared_error(y_test,xgb_pred))
```

```python
# Hyperparameter Tunning for Random forest

import optuna
from sklearn.model_selection import cross_val_score
def objective(trial):
    n_estimators = trial.suggest_int("n_estimators", 10, 500, log=True)
    max_depth = trial.suggest_int("max_depth", 2, 52)
    min_samples_split = trial.suggest_int("min_samples_split", 2, 20)
    min_samples_leaf = trial.suggest_int("min_samples_leaf", 1, 20)


    rf_pipeline = Pipeline(steps=[
          ('columnTransformer',columnTranformer),
          ('rfRegressor',RandomForestRegressor(n_estimators=n_estimators,
          max_depth=max_depth,
          min_samples_split=min_samples_split,
          min_samples_leaf=min_samples_leaf,
          random_state=42))
    ])
    rf_pipeline.fit(x_train, y_train)

    y_pred = rf_pipeline.predict(x_test)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    return mae
```

```python
study = optuna.create_study(direction="minimize",sampler=optuna.samplers.TPESampler())

study.optimize(objective, n_trials=50, show_progress_bar=True)
```

```python
print("Best trial:", study.best_trial)
print("Best hyperparameters:", study.best_params)
```

```python
# Hyper parameter tunning on XGBoost regresor

def objective(trial):
    params = {
        "n_estimators": trial.suggest_int("n_estimators", 100, 3000),
        "max_depth": trial.suggest_int("max_depth", 3, 10),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.3, log=True),
        "subsample": trial.suggest_float("subsample", 0.5, 1.0),
        "colsample_bytree": trial.suggest_float("colsample_bytree", 0.5, 1.0),
        "min_child_weight": trial.suggest_int("min_child_weight", 1, 10),
        "reg_alpha": trial.suggest_float("reg_alpha", 1e-8, 10.0, log=True),
        "reg_lambda": trial.suggest_float("reg_lambda", 1e-8, 10.0, log=True),
        "gamma": trial.suggest_float("gamma", 1e-8, 1.0, log=True),
    }

    xgb_pipeline = Pipeline(steps=[
          ('columnTransformer',columnTranformer),
          ('xgbRegressor',xgb.XGBRegressor(objective='reg:squarederror',
                          random_state=42,**params))
    ])

    xgb_pipeline.fit(x_train, y_train)

    scores = cross_val_score(
        xgb_pipeline, X, y, cv=5,
        scoring="neg_root_mean_squared_error",
        n_jobs=-1
    )

    # Convert negative RMSE back to positive RMSE to minimize it
    rmse = -np.mean(scores)
    return rmse
```

```python
study = optuna.create_study(direction = 'minimize',sampler = optuna.samplers.TPESampler())
study.optimize(objective, n_trials=50, show_progress_bar=True)
```

```python
print(f"Number of finished trials: {len(study.trials)}")
print(f"Best trial score (RMSE): {study.best_trial.value}")
print("Best hyperparameters:")
for key, value in study.best_trial.params.items():
    print(f"  {key}: {value}")
```
